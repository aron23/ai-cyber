# 01. Environment Setup

## Overview
This notebook guides you through setting up the development environment for the Vision Transformer malware detection project. It verifies all dependencies are correctly installed and validates the project structure.

## Learning Objectives
- Verify Python environment and dependencies
- Understand the project directory structure
- Test logging and configuration systems
- Ensure all tools are properly configured

## Prerequisites
- Python 3.12+ installed
- Basic familiarity with virtual environments
- Git installed and configured

## 1. Python Environment

First, let's verify we're running the correct Python version. This project requires Python 3.12 or higher for optimal performance and compatibility with the latest ML libraries.

## 2. Installed Dependencies

Let's verify all required dependencies are installed with compatible versions. These libraries form the foundation of our data processing and machine learning pipeline:

- **pandas & numpy**: Data manipulation and numerical computing
- **PyTorch**: Deep learning framework for Vision Transformers
- **scikit-learn**: Traditional ML algorithms and utilities
- **matplotlib & seaborn**: Data visualization
- **yaml**: Configuration file handling

## 3. Directory Structure

A well-organized directory structure is crucial for collaborative development. Let's explore our project layout:

```
analyst/
├── data/           # Dataset storage
│   ├── raw/        # Original, unprocessed data
│   ├── processed/  # Cleaned and transformed data
│   └── features/   # Engineered features
├── src/            # Source code
├── notebooks/      # Jupyter notebooks
├── tests/          # Unit and integration tests
├── configs/        # Configuration files
└── docs/           # Documentation
```

## 4. Logging Configuration

Proper logging is essential for debugging and monitoring data pipelines. Our logging system:
- Writes to both console and file
- Includes timestamps and log levels
- Supports different verbosity levels
- Helps track data processing progress

## 5. Testing Framework

Quality assurance through automated testing ensures reliable code. Our testing setup includes:
- **pytest**: Modern testing framework
- **Coverage reporting**: Track test completeness
- **Fixtures**: Reusable test data and configurations
- **Integration tests**: Validate end-to-end workflows

## 6. Git Configuration

Version control is set up for collaborative development. Key points:
- Working on branch `group-project`
- Follow conventional commit messages
- Use pull requests for code review
- Maintain clean commit history

1. **Import Errors**: Ensure virtual environment is activated
2. **Version Conflicts**: Check requirements.txt for compatible versions
