# SMS Spam Detection - Text Preprocessing Pipeline

**Author**: Data Scientist  
**Date**: 15/06/2025  
**Task**: DS-002 - Text Preprocessing Pipeline (ACCELERATED)  
**Dataset**: SMS Spam Collection (5,574 messages)  
**Dependencies**: DS-001 EDA insights

## Objectives
1. Implement comprehensive text cleaning pipeline
2. Preserve discriminative features identified in EDA
3. Create modular preprocessing class architecture
4. Validate preprocessing impact on spam patterns
5. Set up stratified data splitting for model training

## Key EDA Insights Driving Preprocessing Strategy
- **Severe class imbalance**: 6.5:1 ratio requires stratified handling
- **Discriminative patterns**: URLs (spam detection), phone numbers, money terms, urgency language
- **Character-level patterns**: Punctuation ratios, uppercase patterns statistically significant
- **Length differences**: Preserve length-based features
- **Vocabulary patterns**: Spam-specific words with 3x+ frequency ratios

## Preprocessing Philosophy
**Balance normalization with discriminative feature preservation**
- Standardize obvious noise (encoding, whitespace)
- Tokenize meaningful patterns (URLs → `<URL>`, phones → `<PHONE>`)
- Preserve statistical patterns (punctuation ratios, case patterns)
- Maintain interpretability for business stakeholders
